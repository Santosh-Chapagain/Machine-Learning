```python
from sklearn.preprocessing import LabelEncoder
lr = LabelEncoder()
```

```python
df['Extracurricular Activities'] = lr.fit_transform(df['Extracurricular Activities'])
```

```python
df.info()
```

```python
x = df.iloc[: , :5]
y = df.iloc[: , -1]
```

```python
x
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
```

```python
clf1 = LinearRegression()
clf2 = DecisionTreeRegressor(max_depth=5)
clf3 = SVR()
```

```python
estimators = [('lr' , clf1) , ('dtr' , clf2) , ('sv' , clf3)]
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import VotingRegressor
```

```python
for estimator in estimators:
    out = cross_val_score(estimator[1] , x , y ,  cv=10 , scoring='r2')
    print(f"{estimator[0]}: {np.round(np.mean(out) ,2)}")
```

```python
vr = VotingRegressor(estimators=estimators)
out = cross_val_score(vr , x , y , cv=10 ,scoring='r2')
print(f"{np.round(np.mean(out) , 2)}")
```
