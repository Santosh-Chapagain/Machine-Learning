## Working with date

```python
date['date'] = pd.to_datetime(date['date'])
```

```python
date.info()
```

```python
#Extract year
date['date_year'] = date['date'].dt.year
date.head()
```

```python
# Extract days
date['date_day'] = date['date'].dt.day
date.head()
```

```python
#day name
date['date_day_name'] = date['date'].dt.day_name()
date.head()
```

```python
# Is weekend or not!!

date['date_is_weekend'] = np.where(date['date_day_name'].isin(['Saturday' , 'Sunday']) , 1 , 0)

date.drop(columns=['product_id' ,'city_id' , 'orders']).head()
```

```python
# Extract week
date['date_week_no'] = date['date'].dt.weekday
date.head()
```

```python
# Extract quarter
date['date_quarter'] = date['date'].dt.quarter
date.head()
```

```python
# Extract semester
date["date_semester"] = np.where(date['date_quarter'].isin([1,2]) , 1 , 2)
date.head()
```

```python
# Times difference with today dates
import datetime
today = datetime.datetime.today()
today
```

```python
today - date['date']
```

```python
# week passed
np.round((today - date['date'])/np.timedelta64(1 , 'W'), 0)
```
