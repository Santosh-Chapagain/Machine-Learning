## Assumption of Linear Regression

1. **Linear Relationship between input and output** .
2. **No Multicolinearity** : Features should be independent . 
3. **Normality of Residual** .
4. **Homoscedasticity**: Homoscedasticity means that the variance (spread) of the errors is constant across all values of the independent variable.
5. **No Autocorrelation of Errors** : It mean when we plot the error then it should not form any pattern . It should be independent .
