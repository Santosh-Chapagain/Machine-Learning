```python
import pandas as pd
import numpy as np 
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

```python
df = pd.read_csv(r"C:\Users\Acer\OneDrive\Desktop\Machine Learning\CSV\Student_Performance.csv")
```

```python
df
```

```python

def clean_data(df):
    
    df['Extracurricular Activities'] = df['Extracurricular Activities'].str.replace("^Yes$", "1", regex=True)
   
    df['Extracurricular Activities'] = df['Extracurricular Activities'].str.replace("^No$", "0", regex=True)
    return df

df_clean = clean_data(df.copy())
df= df_clean.copy()
df
```

```python
df.info()
```

```python
df['Extracurricular Activities'] = df['Extracurricular Activities'].astype(int)
```

```python
x = df.iloc[: , 0:6]
y = df.iloc[: , -1]
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size= 0.2 , random_state=2)
```

```python
x_train.head()
```

```python
from sklearn.linear_model import LinearRegression
lr =LinearRegression()
```

```python
lr.fit(x_train , y_train)
```

```python
y_pred = lr.predict(x_test)
y_pred
```

```python
y_test
```

```python
from sklearn.metrics import accuracy_score , mean_absolute_error , mean_squared_error , r2_score
```

```python
print("Mean Absolute Error:", mean_absolute_error(y_test, y_pred))
print("Mean Squared Error:", mean_squared_error(y_test, y_pred))
print("Root Mean Squared Error:", np.sqrt(mean_squared_error(y_test, y_pred)))
print("R^2 Score:", r2_score(y_test, y_pred))
```
