# 📊 Outliers 

## 🔍 Definition
Outliers are data points that **deviate significantly** from the majority of the dataset.  
They can be **unusually high or low values** that do not follow the general trend.

---

## 🧠 Example
If most students score between **60 and 90**, but one scores **5** or **100**,  
those values are **outliers**.

---

## 📈 Importance of Outliers
- **Affect Averages:** Can distort the mean and standard deviation.  
- **Influence Models:** Can reduce accuracy in statistical or ML models.  
- **Provide Insights:** May indicate rare or important events (e.g., fraud, exceptional cases).

---

## 🧮 Methods to Detect Outliers

### 1. **IQR (Interquartile Range) Method**
```python
Q1 = data.quantile(0.25)
Q3 = data.quantile(0.75)
IQR = Q3 - Q1
lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR
outliers = data[(data < lower_limit) | (data > upper_limit)]


### 2. ***Z_score_method***
from scipy import stats

z_scores = stats.zscore(data)

outliers = data[abs(z_scores) > 3]


### 3. ***Percentile (Quantile) Method***

Covers a fixed percentile range instead of IQR.

lower_limit = data.quantile(0.05)

upper_limit = data.quantile(0.95)

outliers = data[(data < lower_limit) | (data > upper_limit)]

- Any value outside the 5th–95th percentile is an outlier.

- Flexible and suitable for non-normal distributions.
