### Difference between bagging and boosting .

1. In bagging , we use  **Low Bais and High Variance** model but in Boosting , we use **high bais and low variance** model.
2. In bagging , model learn **sequentially**. In boosting , model learn **parallely** .
3. In bagging , all prediction get equal weightage . In boosting, all prediciton get different weightage .
