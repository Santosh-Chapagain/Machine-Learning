# 📌 Naive Bayes for Numerical Datasets – Distribution-Based Notes

---

## 🔹 Key Principle (MOST IMPORTANT ⭐⭐⭐)

> **The choice of Naive Bayes depends on how numerical features are distributed.**

Naive Bayes does **not** work the same way for all numeric data — it assumes a **probability distribution** for features.

---

## 🔹 1️⃣ Gaussian Naive Bayes (GaussianNB)

### 📌 When to use

Use **Gaussian Naive Bayes** when:

* Features are **continuous numerical values**
* Data is **approximately normally distributed** (bell-shaped)

### 📊 Assumed distribution

**Normal (Gaussian) Distribution**

[
P(x|C) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x-\mu)^2}{2\sigma^2} \right)
]

Where:

* ( \mu ) = mean of feature for a class
* ( \sigma^2 ) = variance of feature for a class

### 📌 Examples

* Breast cancer dataset
* Height, weight, age
* Medical measurements
* Sensor readings

### ❌ Do NOT use when

* Features are categorical
* Features are binary (0/1)

---

## 🔹 2️⃣ Multinomial Naive Bayes (MultinomialNB)

### 📌 When to use

Use **Multinomial Naive Bayes** when:

* Features are **counts or frequencies**
* Values are **non-negative integers**

### 📊 Assumed distribution

**Multinomial Distribution**

Used to model:

* Count of events
* Frequency of words

### 📌 Examples

* Bag of Words (text classification)
* TF-IDF features
* Number of clicks

### ❌ Do NOT use when

* Features contain negative values
* Continuous real numbers

---

## 🔹 3️⃣ Bernoulli Naive Bayes (BernoulliNB)

### 📌 When to use

Use **Bernoulli Naive Bayes** when:

* Features are **binary (0/1)**
* Presence / absence matters

### 📊 Assumed distribution

**Bernoulli Distribution**

[
P(x|C) = p^x (1-p)^{(1-x)}
]

### 📌 Examples

* One-Hot Encoded categorical data
* Word present or not
* Yes/No features

---

## 🔹 4️⃣ Poisson Naive Bayes (Conceptual)

### 📌 When to use

* Count-based numeric data
* Rare events per unit time

### 📊 Assumed distribution

**Poisson Distribution**

> ⚠️ Not directly available in sklearn NB, but conceptually important.

### 📌 Examples

* Number of insurance claims per day
* Number of accidents

---

## 🔹 Distribution → Naive Bayes Mapping (EXAM GOLD ⭐⭐⭐)

| Feature Type       | Distribution | Naive Bayes   |
| ------------------ | ------------ | ------------- |
| Continuous numeric | Gaussian     | GaussianNB    |
| Count / frequency  | Multinomial  | MultinomialNB |
| Binary (0/1)       | Bernoulli    | BernoulliNB   |
| Rare event counts  | Poisson      | (Conceptual)  |

---

## 🔹 Common Mistakes ❌

* Using GaussianNB on One-Hot Encoded data
* Using MultinomialNB on continuous features
* Using Label Encoding for categorical features

---

## 🔹 How to decide in practice (Quick checklist ⭐⭐)

Ask yourself:

1. Are values continuous? → **GaussianNB**
2. Are values counts? → **MultinomialNB**
3. Are values binary? → **BernoulliNB**

---

## 🔹 One-Line Ultimate Summary ⭐⭐⭐

> **Naive Bayes choice is driven by feature distribution: Gaussian for continuous data, Multinomial for counts, Bernoulli for binary features.**

```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv(
    r"C:\Users\Acer\OneDrive\Desktop\Machine Learning\CSV\heart.csv")
```

```python
x = df.iloc[: , :-1]
y = df.iloc[: , -1]
```

```python
y
```

```python
x
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x,  y , test_size=0.2 , stratify=y ,random_state=42)
```

```python
x_train
```

```python
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline                  
pipe = Pipeline([('scaler', StandardScaler()), ('gnb', GaussianNB())])

pipe.fit(x_train , y_train)
y_pred = pipe.predict(x_test)
accuracy_score(y_test , y_pred)
```

```python
from sklearn.metrics import recall_score
recall_score(y_test , y_pred)
```
