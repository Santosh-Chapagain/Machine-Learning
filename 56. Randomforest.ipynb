# 🌳 Random Forest — Detailed Notes

## 1. What is Random Forest?
Random Forest is an ensemble learning method that builds multiple decision trees and combines their predictions.  
It is used for:
- Classification
- Regression

It works on the idea that combining many weak/overfitted models can create a strong, stable model.

---

## 2. Why a Single Decision Tree Overfits
A decision tree:
- Learns the training data too deeply  
- Captures noise  
- Has **high variance**  
- Performs differently with small changes in data  

Thus, decision trees are unstable and overfit easily.

---

## 3. How Random Forest Works
### **Step 1: Bootstrapping (Sampling with Replacement)**
Multiple random subsets of data are created from the original dataset.  
Each subset trains one tree.

### **Step 2: Random Feature Selection**
At each node split, only a random subset of features is considered.  
This ensures trees are not similar.

### **Step 3: Aggregation**
- Classification → Majority vote  
- Regression → Average of all tree predictions  

---

## 4. How Random Forest Reduces High Variance (Most Important)

### **A. Bagging reduces instability**
Each tree is trained on different bootstrap samples.  
Because each tree overfits differently, their **average prediction becomes stable**.  
Noise in one tree gets canceled by others.

If we have some outliers in dataset then it will divided into different subset . So each tree trained with small number of outliears whihch also reduce overfit . 

```python
import pandas as pd
import numpy as np
from sklearn.datasets import load_breast_cancer
```

```python
data = load_breast_cancer()
```

```python
x , y = data.data , data.target
```

```python
df = pd.DataFrame(x , y)
```

```python
df.columns = data.feature_names
df['Output'] = data.target
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
x = df.iloc[: , :30]
y = df.iloc[: , -1]
```

```python
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=123)
```

```python
x_train
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(n_estimators=50 , max_features=0.5)
```

```python
rf.fit(x_train , y_train)
```

```python
from sklearn.model_selection import cross_val_score
out = cross_val_score(rf , x_train , y_train , cv=10 , scoring='accuracy')
```

```python
np.mean(out)
```

```python
rf.score(x_test,y_test)
```

```python
from sklearn.metrics import accuracy_score
y_pred = rf.predict(x_test)
accuracy_score(y_test , y_pred)
```

```python
importance_df = pd.DataFrame({
    'Feature': x.columns,
    'Importance': rf.feature_importances_
}).sort_values('Importance', ascending=False)
```
