# Bagging (Bootstrap Aggregating)

## What is Bagging?
Bagging, short for **Bootstrap Aggregating**, is an ensemble learning technique designed to reduce variance and prevent overfitting by training multiple models on different random subsets of the training data.

## How Bagging Works
1. Create multiple **bootstrap samples** by randomly sampling the original dataset with replacement.
2. Train a base estimator (like a decision tree) on each bootstrap sample independently.
3. Combine the predictions from all models:
   - For **classification**, use majority voting.
   - For **regression**, take the average.

## Why Bagging Helps
- **Reduces variance** by averaging out errors from different models.
- **Maintains low bias** by keeping complex base models.
- Works best with **unstable learners** like decision trees that have high variance.

## Advantages
- Decreases overfitting.
- Improves model stability.
- Easy to parallelize training.
- Handles noisy data better.

## Limitations
- Does **not** reduce bias much.
- Computationally expensive with many estimators.
- Less effective for low-variance models (e.g., linear regression).

```python
import pandas as pd
import numpy as np 
```

```python
from sklearn.datasets import load_breast_cancer
data = load_breast_cancer()
x, y = data.data , data.target
```

```python
df = pd.DataFrame(data.data)
```

```python
df.columns = data.feature_names
df['Output'] = data.target
```

```python
df
```

```python
x = df.iloc[: , :30]
y = df.iloc[: , -1]
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
x_train
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import BaggingClassifier
```

```python
bg = BaggingClassifier(estimator=DecisionTreeClassifier())
```

```python
bg.fit(x_train , y_train)
```

```python
y_pred = bg.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test , y_pred)
```

```python
from sklearn.model_selection import cross_val_score
max_samples_list = [0.25 , 0.5, 0.7, 0.9, 1.0]
max_features_list = [0.25 , 0.5, 0.7, 1.0]     

best_score = 0
best_params = {}

for max_samples in max_samples_list:
    for max_features in max_features_list:
        model = BaggingClassifier(
            estimator=DecisionTreeClassifier(),
            n_estimators=50,
            max_samples=max_samples,
            max_features=max_features,
            random_state=42,
            bootstrap=True
        )

        
        scores = cross_val_score(model, x, y, cv=5, scoring='accuracy')
        mean_score = np.mean(scores)

        print(
            f"max_samples={max_samples}, max_features={max_features}, Accuracy={mean_score:.4f}")

        if mean_score > best_score:
            best_score = mean_score
            best_params = {'max_samples': max_samples,
                           'max_features': max_features}

print("\nBest parameters:", best_params)
print("Best accuracy:", best_score)
```

## Pasting

```python
bg.fit(x_train, y_train)
```

```python
y_pred = bg.predict(x_test)
```

```python
accuracy_score(y_test , y_pred)
```

## Random Subspaces

```python
bg = BaggingClassifier(estimator=DecisionTreeClassifier(), bootstrap=True , max_samples=1.0, max_features=0.5 , bootstrap_features=False)
```

```python
bg.fit(x_train , y_train)
y_pred = bg.predict(x_test)
accuracy_score(y_test , y_pred)
```

```python
bg.fit(x_train, y_train)
y_pred = bg.predict(x_test)
accuracy_score(y_test, y_pred)
```
