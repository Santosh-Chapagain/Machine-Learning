# Principal Component Analysis (PCA)

---

## 1. What is PCA?

PCA is a **dimensionality reduction technique** used in data science and machine learning. It transforms a large set of variables into a smaller one while **retaining most of the original information**.

- Think of it as summarizing your data in fewer dimensions without losing much “essence” or variance.
- Useful for **visualization, noise reduction, and speeding up machine learning models**.

---

## 2. Key Idea

PCA finds **new axes (principal components)** in your data:

1. **PC1 (First Principal Component):** Direction along which the data varies the most.
2. **PC2:** Orthogonal to PC1, along which the next most variation exists.
3. And so on…

Each subsequent component explains the **maximum remaining variance** under the constraint of being perpendicular to previous components.

---

## 3. Steps to Perform PCA

1. **Standardize the data**
   - Center the data (subtract mean)
   - Optionally scale it (divide by standard deviation)

2. **Compute the Covariance Matrix**
   - Measures how variables vary together

3. **Compute Eigenvectors and Eigenvalues**
   - Eigenvectors → directions of principal components
   - Eigenvalues → amount of variance captured by each component

4. **Sort Eigenvectors by Eigenvalues**
   - Choose top `k` components that capture most variance

5. **Transform the Data**
   - Project original data onto these top `k` components

---

## 4. Why PCA is Useful

- **Reduces complexity** → fewer features
- **Removes noise** → ignores insignificant variance
- **Visualization** → 2D or 3D plots of high-dimensional data
- **Speeds up ML models** → fewer features = faster computation

---

## 5. Example

Suppose you have a dataset with **5 features**. After PCA:

- PC1 → captures 70% variance
- PC2 → captures 20% variance
- PC3 → captures 5% variance
- Remaining PCs → only 5% variance

If your goal is dimensionality reduction, you can use **just PC1 and PC2** and still retain 90% of the data’s information.

---

## 6. Things to Keep in Mind

- PCA is **unsupervised** → doesn’t consider target labels.
- Works best with **linear relationships**.
- Can be sensitive to **scaling**; always standardize numeric features.
