## 🔹 What is Naive Bayes?

**Naive Bayes** is a **supervised machine learning algorithm** based on **Bayes’ Theorem**.

It is mainly used for:

* **Classification problems**
* Especially **text classification**

> 🎯 **Main Idea:** Predict the class with the **highest probability** given the input features.

---

## 🔹 Why is it called “Naive”?

Because it assumes:

> ❗ **All features are conditionally independent given the class**

Even though this assumption is often false in real data, Naive Bayes still works surprisingly well.

---

## 🔹 Bayes’ Theorem (Core Formula ⭐⭐⭐)

[
P(C|X) = \frac{P(X|C) \cdot P(C)}{P(X)}
]

Where:

* **C** → class (label)
* **X** → input features
* **P(C|X)** → Posterior probability
* **P(X|C)** → Likelihood
* **P(C)** → Prior probability
* **P(X)** → Evidence (constant)

---

## 🔹 Naive Independence Assumption ⭐⭐

For multiple features:
[
P(X|C) = P(x_1|C) \cdot P(x_2|C) \cdot ... \cdot P(x_n|C)
]

This simplifies computation drastically.

---

## 🔹 Final Classification Rule (VERY IMPORTANT ⭐⭐⭐)

[
\hat{C} = \arg\max_C ; P(C) \prod_{i=1}^{n} P(x_i | C)
]

👉 **Choose the class with maximum posterior probability.**

---

## 🔹 Types of Naive Bayes

### 1️⃣ Gaussian Naive Bayes

Used when features are **continuous** and normally distributed.

**Likelihood formula:**
[
P(x|C) = \frac{1}{\sqrt{2\pi\sigma_C^2}} \exp\left( -\frac{(x-\mu_C)^2}{2\sigma_C^2} \right)
]

---

### 2️⃣ Multinomial Naive Bayes

Used for:

* **Text data**
* **Word counts / frequencies**

Common in:

* Spam detection
* Sentiment analysis

---

### 3️⃣ Bernoulli Naive Bayes

Used when features are:

* **Binary (0/1)**

Example:

* Word present or not present

---

## 🔹 Log Probability Trick ⭐⭐

To avoid numerical underflow:

[
\log P(C|X) = \log P(C) + \sum_{i=1}^{n} \log P(x_i|C)
]

Used in all practical implementations.

---

## 🔹 Advantages of Naive Bayes

✅ Very fast training and prediction
✅ Works well with high-dimensional data
✅ Performs well on small datasets
✅ Excellent for text classification

---

## 🔹 Disadvantages of Naive Bayes

❌ Strong independence assumption
❌ Poor performance when features are highly correlated

---

## 🔹 When to Use Naive Bayes

✔ Spam detection
✔ Sentiment analysis
✔ Document classification
✔ Medical diagnosis (baseline model)

---

## 🔹 What You DON’T Need to Remember ❌

* Proof of Bayes’ theorem
* Mathematical derivation
* Measure-theoretic probability

👉 **Final formulas + intuition are enough**.

---

## 🔹 One-Line Ultimate Summary ⭐⭐⭐

> **Naive Bayes is a probabilistic classifier that applies Bayes’ theorem with a strong independence assumption between features.**
