# Gradient Boosting Classifier — Internal Working 

---

## 1. What is Gradient Boosting Classifier?

Gradient Boosting Classifier is an **ensemble learning algorithm** that builds
multiple **weak learners (decision trees)** sequentially.

Each new tree is trained to **correct the mistakes** made by the previous model
by minimizing a **loss function** using **gradient descent principles**.

---

## 2. Key Concepts Used Internally

- Additive model
- Gradient descent
- Loss function (Log Loss for classification)
- Residuals (negative gradients)
- Weak learners (shallow trees)

---

## 3. Loss Function Used (Binary Classification)

Gradient Boosting Classifier uses **Log Loss (Binary Cross-Entropy)**:

$$
L(y, p) = -[y \log(p) + (1 - y)\log(1 - p)]
$$

Where:
- `y` = actual label (0 or 1)
- `p` = predicted probability

---

## 4. Overall Workflow (High Level)

1. Start with an initial constant model
2. Compute errors (residuals / gradients)
3. Train a decision tree on residuals
4. Update the model using learning rate
5. Repeat steps 2–4
6. Make final prediction using all trees

---

## 5. Step-by-Step Internal Working

---

### Step 1: Initialize the Model (f₀)

For classification, the initial prediction is **log-odds** of positive class:

$$
f_0 = \log \left( \frac{p}{1 - p} \right)
$$

Where:
$$
p = \frac{\text{number of positive samples}}{\text{total samples}}
$$

This is a **constant value**, not a tree.

---

### Step 2: Convert Log-Odds to Probability

Using sigmoid function:

$$
p = \frac{1}{1 + e^{-f(x)}}
$$

---

### Step 3: Compute Residuals (Negative Gradient)

Residuals are **NOT simple (y - y_pred)**.

They are computed as the **negative gradient of loss**:

$$
r_i = y_i - p_i
$$

Where:
- `y_i` = actual label
- `p_i` = predicted probability

These residuals represent **how much correction is needed**.

---

### Step 4: Train Decision Tree on Residuals

- Input: Feature matrix `X`
- Target: Residuals `r_i`
- Tree depth is kept **small (1–3)**

The tree learns:
> "For this region of feature space, how should prediction be adjusted?"

---

### Step 5: Compute Leaf Values (Gamma)

Each leaf predicts a value `γ` that minimizes the loss:

$$
\gamma = \frac{\sum r_i}{\sum p_i (1 - p_i)}
$$

This step ensures **optimal correction** per leaf.

---

### Step 6: Update the Model

Model update rule:

$$
f_{m}(x) = f_{m-1}(x) + \eta \cdot \gamma
$$

Where:
- `η` = learning rate
- `γ` = leaf value from tree

This is why gradient boosting is called an **additive model**.

---

### Step 7: Repeat for M Trees

Repeat:
- Compute probabilities
- Compute residuals
- Train tree
- Update model

Until:
- `n_estimators` reached
- OR improvement becomes very small

---

## 6. Final Prediction

After all trees:

1. Sum predictions from all trees:
$$
f(x) = f_0 + \sum \eta f_m(x)
$$

2. Convert to probability using sigmoid
3. Apply threshold (default = 0.5)

---

## 7. Stopping Criteria (Manual Implementation)

Stop when:
- Residuals are very small
- Loss stops decreasing
- Validation performance stops improving

In sklearn:
- Controlled using `n_estimators`
- Or `early_stopping`

---

## 8. Important Hyperparameters (Internal Impact)

| Parameter | Internal Effect |
|--------|---------------|
learning_rate | Controls step size of updates |
n_estimators | Number of boosting steps |
max_depth | Controls complexity of trees |
subsample | Adds randomness to reduce overfitting |

---

## 9. Why Shallow Trees Are Used?

- Prevent overfitting
- Each tree acts as a **weak learner**
- Boosting works best with weak models

---

## 10. Comparison with Random Forest

| Gradient Boosting | Random Forest |
|-----------------|---------------|
Sequential learning | Parallel learning |
Focuses on mistakes | Independent trees |
High accuracy | Faster training |
Sensitive to noise | More robust |

---

## 11. Interview One-Liner (IMPORTANT)

"Gradient Boosting Classifier builds decision trees sequentially, where each tree
is trained on the negative gradient of the loss function to correct errors made
by the previous model."

---

## 12. Key Takeaway

- Gradient Boosting = Gradient Descent + Trees
- Residuals = Direction to reduce loss
- Learning rate controls stability
- Trees correct errors step-by-step
