# Voting Ensemble

## What is Voting Ensemble?
Voting ensemble is an ensemble learning technique that combines predictions from multiple different models to make a final prediction.

- For **classification**, the final class is predicted by **majority voting** (hard voting) or by averaging predicted probabilities (soft voting).
- For **regression**, predictions are usually averaged.

## Types of Voting
- **Hard Voting:** Each model votes for a class label, and the class with the most votes wins.
- **Soft Voting:** Each model predicts class probabilities, and the average probability is used to decide the final class.

## Why Use Voting Ensemble?
- Combines strengths of different models.
- Can improve performance over individual models.
- Reduces risk of picking a single poor-performing model.
- Useful when you have diverse models (e.g., Logistic Regression, Decision Trees, SVM).

```python
import pandas as pd
import numpy as np 
```

```python
df = pd.read_csv(
    r"C:\Users\Acer\OneDrive\Desktop\Machine Learning\CSV\heart.csv")
```

```python
df.head()
```

```python
x = df.iloc[: , 0:13]
y = df.iloc[: , -1]
```

```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(x , y , test_size=0.2 , random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

```python
clf1 = LogisticRegression(max_iter=1500)
clf2 = DecisionTreeClassifier()
clf3 = RandomForestClassifier()
```

```python
estimators = [('lr' , clf1) , ('dc' , clf2) , ('rf' , clf3)]
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.metrics import accuracy_score
```

```python
for estimator in estimators:
    x = cross_val_score(estimator[1], x_train, y_train, cv=10, scoring='accuracy')
    print(estimator[0] , np.round(np.mean(x) , 2))
```

```python
from sklearn.ensemble import VotingClassifier
```

## Hard voting

```python
vc= VotingClassifier(estimators=estimators)

x = cross_val_score(vc , x_train, y_train , cv=10 , scoring='accuracy')
print(np.round(np.mean(x) , 2))
```

## Soft voting

```python
vc2 = VotingClassifier(estimators=estimators , voting='soft')

x = cross_val_score(vc2, x_train, y_train, cv=10, scoring='accuracy')
print(np.round(np.mean(x), 2))
```
