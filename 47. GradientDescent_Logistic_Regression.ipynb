```python
from sklearn.datasets import make_classification

x, y = make_classification(
    n_samples=100,
    n_features=2,
    n_redundant=0,
    n_informative=1,
    n_classes=2,
    n_clusters_per_class=1,
    class_sep=30.0,    # makes data clearly separable
    random_state=41,
    hypercube=False
)
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[:, 0], x[:, 1], c=y, cmap='winter',  s=100)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr= LogisticRegression(penalty=None ,solver="lbfgs", max_iter=10000 )
```

```python
lr.fit(x, y)
```

```python
m = -(lr.coef_[0][0]/lr.coef_[0][1])
b = -(lr.intercept_/lr.coef_[0][1])
```

```python
import numpy as np 
x_input = np.linspace(-3, 3, 100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(8, 4))
plt.plot(x_input, y_input, color='red', linewidth=3)
plt.scatter(x[:, 0], x[:, 1], c=y, cmap='winter', s=100)
plt.title("Using SKlearn")
plt.ylim(-3, 2)
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
```

```python
def get(x, y):
    x = np.insert(x, 0, 1, axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.5

    for i in range(10000):
        
        y_hat = sigmoid(np.dot(x, weights))
        weights = weights + lr * np.dot((y - y_hat) , x)/x.shape[0]

    return weights[1:], weights[0]
```

```python
coef_ , intercept_ = get(x, y)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3, 3, 100)
y_input1 = m*x_input + b
```

```python
plt.figure(figsize=(8, 4))
plt.plot(x_input1, y_input1, color='brown', linewidth=3)
plt.plot(x_input, y_input, color='red', linewidth=3)
plt.scatter(x[:, 0], x[:, 1], c=y, cmap='winter', s=100)
plt.title("Using own code")
plt.ylim(-3, 2)
```
