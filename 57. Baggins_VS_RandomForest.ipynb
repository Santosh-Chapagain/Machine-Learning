# Difference Between Bagging and Random Forest

## 1. Base Estimator (Algorithms Used)

### Bagging
- Bagging is a general ensemble technique.
- It can use **multiple different algorithms** as the base estimator:
  - Decision Tree
  - KNN
  - SVM
  - Logistic Regression
  - etc.
- In bagging, you are free to choose any model.

### Random Forest
- Random Forest is **not general**.
- It can use **only Decision Trees** as base estimators.
- It is basically a collection of many randomized decision trees.

---

## 2. Feature Sampling Behavior

### Bagging (with max_features)
- If you set `max_features = 2`, bagging selects **a fixed set of 2 features** for the **entire tree**.
- For example:
  - If it selects features **(1, 2)**,  
    then **every split** inside that tree uses only feature 1 and feature 2.
- Feature subset is **constant** throughout the whole tree.

### Random Forest
- If you set `max_features = 2`, Random Forest selects **2 random features at every split**.
- Example within the same tree:
  - Split 1 → (1, 2)
  - Split 2 → (2, 4)
  - Split 3 → (3, 5)
- Feature subset is **dynamic** and changes at every node.

---

## ⭐ Summary

1. **Bagging can use any algorithm (DT, KNN, SVM, etc.), while Random Forest only uses Decision Trees.**

2. **Bagging’s feature sampling is fixed for the whole tree, whereas Random Forest randomly selects new features at every split.**
